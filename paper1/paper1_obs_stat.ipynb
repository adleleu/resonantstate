Plot the observed population in the $\Delta_{MMR},e_r$ plane, with the color & transparency depending on the robustnes & precision of the TTV solution

```python
import pandas as pd
import matplotlib.pyplot as plt 
from f_plots import *
from resonantstate.ell2SFM import *


case='paper1_obs_selection'
df = pd.read_parquet('data/'+case+'_rs')

#DMMR vs Xres
fig, ax = plt.subplots(1, 1, figsize=(5,5))


plt_rob(ax,df['DMMR'],df['er'],df['DMMR_e1'],df['DMMR_e2'],df['er_e1'],df['er_e2'],df['author_name'],df['reason'],alpha_text=0)


er_lim=(-4,10)
DMMR_lim=(-60,60)

plot_topology_DMMR(ax, DMMR_lim, er_lim,dark=False,linewidth=1,legend=True)
plt.legend()
```

Same but adding the color background

```python
fig, ax1 = plt.subplots(1, 1, figsize=(5,5))


plt_rob(ax1,df['DMMR'],df['er'],df['DMMR_e1'],df['DMMR_e2'],df['er_e1'],df['er_e2'],df['author_name'],df['reason'],alpha_text=0)


er_lim=(-4,10)
DMMR_lim=(-60,60)
ax1=theory_background(ax1,DMMR_lim,er_lim)


plot_topology_DMMR(ax1, DMMR_lim, er_lim,dark=False,linewidth=1,legend=False)
plt.legend(loc='upper left')
```
